# **Handling Missing Values By Mean,Median and Mode**

```python
#filling every missing value with the word "missing value"
df_pyspark.na.fill('Missing Value').show()
```

```python
#Since age,experience and salary are integers they wont be filled with the name missing value so we take this approach
df_pyspark.na.fill({
    "Name": "Missing Value",
    "age": 0,
    "Experience": 0,
    "Salary": 0
}).show()
```

```python
df_pyspark.show()
```

```python
# Import the Imputer class from PySpark's Machine Learning (ML) library.
# This tool is specifically designed to handle missing (null) values in numerical columns.
from pyspark.ml.feature import Imputer

# Initialize the Imputer estimator (it's best practice to use lowercase 'imputer' for the variable name
# so we don't overwrite the 'Imputer' class itself!)
imputer = Imputer(
    # 1. Specify the target columns containing missing values (Nulls/NaNs) that need replacing.
    inputCols=['age', 'Experience', 'Salary'],

    # 2. Define the names of the new output columns that will hold the cleaned data.
    # This list comprehension dynamically generates: ['age_inputed', 'Experience_inputed', 'Salary_inputed']
    outputCols=["{}_inputed".format(c) for c in ['age', 'Experience', 'Salary']]
).setStrategy("mean")  # 3. Set the math strategy to calculate the "mean" (average) of each column to fill the nulls.
                       # Other options include "median" or "mode".


# This single line performs three major steps under the hood:
#
# Step A: .fit(df_pyspark)
#   Calculates the actual mean (average) values for 'age', 'Experience', and 'Salary'
#   from your non-null rows. It creates a "fitted model" containing these values.
#
# Step B: .transform(df_pyspark)
#   Applies those calculated averages to fill the missing spaces in your DataFrame,
#   and outputs the new columns.
#
# Step C: .show()
#   An action that triggers the computations and immediately displays the resulting
#   DataFrame with the new imputed columns.
imputer.fit(df_pyspark).transform(df_pyspark).show()
```

```python
#in pyspark we dont do train test split we 1st create a vector assembler(which is a group of our independent features) and then split the data
#[Age,Experience]------------>new feature-------->independent feature (To do this we use vectorassembler)
```

```python
# Import the VectorAssembler class from PySpark's Machine Learning library.
# Machine learning algorithms in PySpark require all input features to be combined
# into a single array (vector) column, and this tool is built specifically for that job.
from pyspark.ml.feature import VectorAssembler

# Initialize the VectorAssembler transformer.
featureassembler = VectorAssembler(
    # 1. Specify the list of individual columns you want to use as predictors (features).
    inputCols=["age", "Experience"],

    # 2. Name the single new column that will hold the merged feature vector.
    # In PySpark, this is standard practice before feeding data into an ML model.
    outputCol="Independent Features"
)
```

```python
# Apply the VectorAssembler to your dataset (called 'training' here).
# This takes the individual columns specified in the assembler ('age' and 'Experience')
# and groups them row-by-row into a brand-new vector column named 'Independent Features'.
output = featureassembler.transform(training)
```

```python
output.show()
```

```python
output.columns
```

```python
# Create a streamlined DataFrame containing ONLY the two columns needed for modeling:
# 1. 'Independent Features' -> Your input variables (X) packaged as a vector.
# 2. 'Salary' -> Your target/dependent variable (Y) that you want to predict.
finalized_data = output.select('Independent Features', 'Salary')

# An action that executes the column selection and prints the trimmed dataset to the console.
finalized_data.show()
```

```python
# Import the LinearRegression class from PySpark's Machine Learning library.
# This algorithm will try to find a line of best fit to predict 'Salary'
# based on your 'Independent Features' (age and experience).
from pyspark.ml.regression import LinearRegression

# Split your finalized dataset into two distinct dataframes:
# 1. train_data -> gets roughly 75% of the rows (used to teach the model).
# 2. test_data -> gets roughly 25% of the rows (held back to evaluate the model later).
# Note: This split is randomized. If you want reproducible splits, add a seed (e.g., randomSplit([0.75, 0.25], seed=42)).
train_data, test_data = finalized_data.randomSplit([0.75, 0.25])

# Initialize the Linear Regression estimator configuration.
# Here we explicitly map out the inputs and outputs:
# - featuresCol tells Spark where to find the packed independent input vector.
# - labelCol tells Spark which column holds the true answer we want to predict.
regressor = LinearRegression(featuresCol='Independent Features', labelCol='Salary')

# Train (fit) the model using your 75% training dataset.
# Under the hood, this triggers a Spark Action. Spark runs mathematical equations (like Ordinary Least Squares)
# across your cluster to calculate the optimal weights (coefficients) and intercept for your data.
# The resulting variable is a LinearRegressionModel object that can make predictions.
regressor = regressor.fit(train_data)
```

```python
# Access the calculated weights (slopes) assigned to each of your input features.
# This returns a PySpark DenseVector containing one numerical value for every
# feature in your 'Independent Features' vector (in this case: age and Experience).
regressor.coefficients
```

```python
# Access the calculated Y-intercept (the baseline value) of the linear regression equation.
# This returns a single float representing the starting salary prediction when
# all independent features (age and experience) are equal to zero.
regressor.intercept
```

```python
# Pass your unseen testing data into the trained model's .evaluate() method.
# This runs your test_data through the regression equation and calculates key
# performance metrics (like R-squared, MAE, and RMSE) by comparing
# the predicted salaries against the actual salaries in test_data.
pred_results = regressor.evaluate(test_data)
```

**What MAE and MSE Actually Do**

Both metrics measure prediction error—the gap between your model's salary guess ($\hat{y}$) and the actual true salary ($y$). However, they weigh those errors differently.

Actual Salary:  $100,000

    Predicted Salary:  $105,000
---------------------------------

          Error (Gap):   $5,000

 Absolute Error (|x|):   $5,000

   Squared Error (x²):   25,000,000


**1. Mean Absolute Error (MAE)What it does:**

Takes the absolute difference for every row (ignoring whether it was an over-guess or under-guess) and averages them together.Unit of measurement: Stays in the exact same units as your target variable (dollars).Plain English Meaning: "On average, our model's predictions are off by $X dollars."

**2. Mean Squared Error (MSE)What it does:**

 Takes each error gap, squares it, and then averages those squared values.Unit of measurement: Measured in squared units (e.g., $\text{dollars}^2$), which makes it hard to read intuitively.Plain English Meaning: Because it squares the errors before averaging, large mistakes get exponentially magnified. An error of $\$10,000$ contributes $100,000,000$ to the total, while an error of $\$1,000$ contributes only $1,000,000$.
