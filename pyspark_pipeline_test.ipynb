```python
# Group the dataset by the 'sex' column and calculate the sum total of all 'tip' values for each gender.
# .groupBy('sex') collects all matching rows for each unique value ('Male', 'Female'),
# .sum('tip') aggregates the tip amounts, and .show() prints the final summary table.
df_tips.groupBy('sex').sum('tip').show()
```

```python
# Group the DataFrame by the 'sex' column and pass a dictionary to .agg() 
# to calculate the total sum of the 'tip' column.
# Format: {'column_name': 'aggregate_function'}
df_tips.groupBy('sex').agg({'tip': 'sum'}).show()
```

```python
# Group by 'sex', sum the 'tip' values using dictionary aggregation, 
# and then rename the auto-generated 'sum(tip)' column to 'total_bill'.
df_tips.groupBy('sex').agg({'tip':'sum'}).withColumnRenamed('sum(tip)','total_bill').show()
```

```python
# Group by 'sex', sum the 'tip' values, and then chain two .withColumnRenamed() calls:
# 1. Renames the auto-generated 'sum(tip)' column to 'total_bill'
# 2. Renames the 'sex' column to 'gender'
df_tips.groupBy('sex').agg({'tip':'sum'}).withColumnRenamed('sum(tip)','total_bill').withColumnRenamed('sex','gender').show()
```

```python
# Group the dataset by 'sex', sum the tips, rename columns, and add 10 to total_bill:
# 1. .groupBy('sex'): Groups rows by gender ('Male', 'Female').
# 2. .agg({'tip':'sum'}): Sums tips, creating a column named 'sum(tip)'.
# 3. .withColumnRenamed('sum(tip)','total_bill'): Renames 'sum(tip)' to 'total_bill'.
# 4. .withColumnRenamed('sex','gender'): Renames 'sex' to 'gender'.
# 5. .withColumn('total_bill', ... + 10): Adds 10 to the total_bill column.
# 6. .show(): Triggers execution and prints the resulting DataFrame.

df_tips.groupBy('sex').agg({'tip':'sum'}).withColumnRenamed('sum(tip)','total_bill').withColumnRenamed('sex','gender').withColumn('total_bill',df_tips['total_bill']+10).show()
```

```python
# Group the dataset by 'sex', calculate total tips, and display the result:
# 1. .groupBy('sex'): Collects all rows corresponding to each unique value in the 'sex' column ('Male', 'Female').
# 2. .agg({'tip':'sum'}): Uses dictionary aggregation to compute the sum of the 'tip' column, producing an output column named 'sum(tip)'.
# 3. .show(): Executes the Spark action and prints the resulting summary table to the screen.
df_tips.groupBy('sex').agg({'tip':'sum'}).show()
```

**One-Hot Encoding**

What it does: Creates a new binary (0 or 1) column for every single unique category in a feature.
How it looks: Red, Green, and Blue become three columns (is_red, is_green, is_blue). Red is [1, 0, 0].
When to use: Use for unordered (nominal) data like colors, cities, or animal types, where no option is greater than another.
Main drawback: Can cause a massive increase in columns if a feature has hundreds of unique values.

**Ordinal Encoding**

What it does: Assigns a single integer to each category based on a specific, logical rank or scale.
How it looks: Low, Medium, and High become one column with values 1, 2, 3.
When to use: Use for ordered data where the rank matters, such as education level (High School < College < Degree) or shirt sizes.

One-hot encoding turns each category into its own separate binary column (0 or 1), while ordinal encoding replaces ordered categories with a single column of sequential integers 

| Feature | Ordinal Encoding | One-Hot Encoding |
| :--- | :--- | :--- |
| **Best for** | Ordered categories (*Education level, Rating, Size*) | Unordered categories (*Gender, Country, Color*) |
| **Result** | Single column with numbers | Multiple columns with 0s and 1s |
| **PySpark Tool** | `StringIndexer` | `OneHotEncoder` |

```python
# 'sex', 'smoker', 'day', 'time' are categorical features and need to be converted to numerical values for ML algorithms.

# Import StringIndexer from PySpark's Machine Learning Feature library.
# StringIndexer maps string categories to numerical indices ordered by label frequency (most frequent category gets 0.0).
from pyspark.ml.feature import StringIndexer

# Initialize StringIndexer for a single categorical column ('sex'):
# - inputCol: The existing string column to encode ('sex').
# - outputCol: The new column name that will store the encoded numerical values ('sex_indexed').
indexer = StringIndexer(inputCol='sex', outputCol='sex_indexed')

# Fit the indexer to learn the category frequencies, then transform the DataFrame to append the new numerical column.
df_r = indexer.fit(df_tips).transform(df_tips)

# Display the resulting DataFrame showing all original columns plus 'sex_indexed'.
df_r.show()
```

```python
# Perform multi-column ordinal encoding using inputCols and outputCols.
from pyspark.ml.feature import StringIndexer

# Initialize StringIndexer with lists of column names:
# - inputCols: List of original categorical string columns to encode.
# - outputCols: List of corresponding new column names for the numerical outputs.
# (Fixed: Added missing closing quote on 'time')
indexer = StringIndexer(
    inputCols=['sex','smoker', 'day', 'time'], 
    outputCols=['sex_indexed','smoker_indexed', 'day_indexed', 'time_indexed']
)

# .fit() calculates category frequencies across the dataset for each column.
# .transform() appends the 3 new indexed columns to the DataFrame.
df_r = indexer.fit(df_tips).transform(df_tips)

# Display the updated DataFrame containing the original features alongside their new indexed versions.
df_r.show()
```

```python
df_r.columns
```

```python
# Import VectorAssembler from PySpark's feature module.
from pyspark.ml.feature import VectorAssembler

# Step 1: Combine independent features into a single feature vector.
# Note: Replaced 'tip' with 'total_bill' so you aren't predicting 'tip' using 'tip'!
featureassembler = VectorAssembler(
    inputCols=['size', 'sex_indexed', 'smoker_indexed', 'day_indexed', 'time_indexed'], 
    outputCol='independent_features'
)

# Transform the DataFrame and store it in 'output'
output = featureassembler.transform(df_r)
output.show()
```

```python
output.select('independent_features').show()
```

```python
finalizedData = output.select('independent_features','total_bill')
finalizedData.show()
```

```python
# Step 2: Initialize and train the Linear Regression model.
from pyspark.ml.regression import LinearRegression

#train test split
train_data,test_data = finalizedData.randomSplit([0.75,0.25])


regressor = LinearRegression(featuresCol='independent_features', labelCol='total_bill')

regressor = regressor.fit(train_data)
```

```python
# Step 3: Inspect the trained model's parameters and evaluation metrics.

# Coefficients (weights assigned to each feature)
print("Coefficients:", regressor.coefficients)
# Intercept (baseline target value)
print("Intercept:", regressor.intercept)


# Root Mean Squared Error (RMSE)
print("RMSE:", regressor.summary.rootMeanSquaredError)


# R-Squared (R² Score)
print("R² Score:", regressor.summary.r2)


```

```python
#predictions 
pred_results=regressor.evaluate(test_data)
pred_results.predictions.show()
```
